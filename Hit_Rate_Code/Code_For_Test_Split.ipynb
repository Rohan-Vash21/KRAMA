```python
import pandas as pd

LANGUAGE = "Choose your desired language"
TEST_PATHS = {
    "Llama-3.2-1B-Instruct": f"Final_output_test_Llama1Bit_file_path",
    "Sarvam-1": f"Final_output_test_Sarvam_file_path",
    "Gemma-2-2b-it": f"Final_output_test_Gemma2Bit_file_path",
    "Llama-3.2-3B-Instruct": f"Final_output_test_Llama3Bit_file_path",
    "Aya-23-8B": f"Final_output_test_Aya8B_file_path",
    "Llama-3.1-8B-Instruct": f"Final_output_test_Llama8Bit_file_path"
}

def run_test_diagnostics():
    print(f"--- TEST SPLIT DIAGNOSTICS: {LANGUAGE.upper()} ---")
    results = []

    for model_name, path in TEST_PATHS.items():
        try:
            test_df = pd.read_csv(path)
            total_test = len(test_df)

            if 'Final_Voted_Pred' in test_df.columns and 'RAG_Pred' in test_df.columns:
                actual_overturns_test = len(test_df[test_df['Final_Voted_Pred'] != test_df['RAG_Pred']])

                results.append({
                    "Model": model_name,
                    "Total": total_test,
                    "Overturns": f"{actual_overturns_test} ({round((actual_overturns_test/total_test)*100, 2)}%)"
                })
            else:
                 print(f"[{model_name}] Columns 'Final_Voted_Pred' or 'RAG_Pred' missing in CSV.")

        except FileNotFoundError:
            print(f"[{model_name}] File not found at {path}. Please verify the path.")

    if results:
        print("\n==========================================================")
        report_df = pd.DataFrame(results)
        print(report_df.to_string(index=False))
        print("==========================================================")

run_test_diagnostics()
```
